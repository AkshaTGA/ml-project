```python
# ============================================================
# COMPLETE TRAINING + INFERENCE SCRIPT (ONE BLOCK)
# ============================================================

import os
import torch
import joblib
import xgboost as xgb
import pandas as pd
import numpy as np

from PIL import Image
from torch.utils.data import Dataset, DataLoader
from torchvision import models, transforms

# ============================================================
# CONFIG
# ============================================================

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
IMG_ROOT = "/kaggle/input/csiro-biomass/"
TRAIN_CSV = "/kaggle/input/csiro-biomass/train.csv"
TEST_CSV  = "/kaggle/input/csiro-biomass/test.csv"

# ============================================================
# CNN BACKBONE (EMBEDDING MODEL)
# ============================================================

class ResNetBackbone(torch.nn.Module):
    def __init__(self):
        super().__init__()
        model = models.resnet50(pretrained=True)
        model.fc = torch.nn.Identity()
        self.model = model

    def forward(self, x):
        return self.model(x)

cnn_model = torch.load("/kaggle/input/embedding/pytorch/default/1/backbone_full.pt", map_location=DEVICE, weights_only= False)
cnn_model.eval()


# ============================================================
# IMAGE DATASET
# ============================================================

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

class ImgDataset(Dataset):
    def __init__(self, df):
        self.df = df.reset_index(drop=True)

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        img = Image.open(
            os.path.join(IMG_ROOT, self.df.loc[idx, "image_path"])
        ).convert("RGB")
        return transform(img)

# ============================================================
# EMBEDDING EXTRACTION
# ============================================================

def extract_embeddings(df, batch_size=32):
    loader = DataLoader(
        ImgDataset(df),
        batch_size=batch_size,
        shuffle=False
    )

    feats = []
    with torch.no_grad():
        for x in loader:
            x = x.to(DEVICE)
            feats.append(cnn_model(x).cpu().numpy())

    return np.vstack(feats)

# ============================================================
# TRAINING (RUN ONCE – COMMENT OUT FOR SUBMISSION)
# ============================================================

train_df = pd.read_csv(TRAIN_CSV)

# Extract embeddings once
train_embeddings = extract_embeddings(train_df)

# Split into 4 datasets based on target_name
target_groups = train_df["target_name"].unique()[:5]
joblib.dump(target_groups.tolist(), "target_groups.pkl")

for g in target_groups:
    idx = train_df["target_name"] == g

    X = train_embeddings[idx]
    y = train_df.loc[idx, "target"].values

    model = xgb.XGBRegressor(
        n_estimators=500,
        max_depth=8,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        objective="reg:squarederror",
        n_jobs=-1
    )

    model.fit(X, y)
    model.save_model(f"xgb_{g}.json")

print("✅ Training completed")

# ============================================================
# INFERENCE (USED BY KAGGLE RERUN)
# ============================================================

# Load models
target_groups = joblib.load("target_groups.pkl")

xgb_models = {}
for g in target_groups:
    m = xgb.XGBRegressor()
    m.load_model(f"xgb_{g}.json")
    xgb_models[g] = m

# Embed single image
def embed_single(path):
    img = Image.open(path).convert("RGB")
    x = transform(img).unsqueeze(0).to(DEVICE)
    with torch.no_grad():
        return cnn_model(x).cpu().numpy()

# Predict
test_df = pd.read_csv(TEST_CSV)
preds = []

for _, row in test_df.iterrows():
    emb = embed_single(
        os.path.join(IMG_ROOT, row["image_path"])
    )

    model = xgb_models[row["target_name"]]
    preds.append(model.predict(emb)[0])

# Submission
submission = pd.DataFrame({
    "sample_id": test_df["sample_id"],
    "target": preds
})

submission.to_csv("submission.csv", index=False)
print("📤 submission.csv generated")
```
